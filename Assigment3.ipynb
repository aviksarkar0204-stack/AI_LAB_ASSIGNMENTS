```python
import seaborn as sns
iris = sns.load_dataset("iris")
```

```python
iris
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

print(iris.head())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
iris["species"] = le.fit_transform(iris["species"])
```

```python
iris.head()
```

```python
X = iris.drop(columns=['species'])
y = iris['species']
```

```python
X_train, X_test , y_train, y_test = train_test_split(X,y,shuffle=True,test_size=0.2,random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
model = KNeighborsClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=200),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Random Forest": RandomForestClassifier()
}
for name,model in models.items():
  model.fit(X_train,y_train)
  y_pred = model.predict(X_test)
  print(name)
  print("Accuracy:",accuracy_score(y_test,y_pred))
  print("\nClassification Report:\n",classification_report(y_test,y_pred))
  print("\nConfusion Matrix:\n",confusion_matrix(y_test,y_pred))
  print("-"*50)
```

```python
# Example Graph (Adjacency List Representation)
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

# Depth-First Search (DFS) implementation
def dfs(graph, start_node, visited=None):
    if visited is None:
        visited = set()
    visited.add(start_node)
    print(start_node, end=' ')
    for neighbor in graph.get(start_node, []):
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

# Breadth-First Search (BFS) implementation
def bfs(graph, start_node):
    visited = set()
    queue = [start_node]
    visited.add(start_node)

    while queue:
        node = queue.pop(0) # Dequeue
        print(node, end=' ')

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor) # Enqueue

print("DFS Traversal starting from 'A':")
dfs(graph, 'A')
print("\n\nBFS Traversal starting from 'A':")
bfs(graph, 'A')
```

```python
from collections import deque
import matplotlib.pyplot as plt
import networkx as nx
from matplotlib.animation import FuncAnimation

# 1. Initialize Graph
G = nx.Graph()
edge = {
    ("A", "B"),
    ("A", "C"),
    ("B", "D"),
    ("B", "E"),
    ("C", "F"),
    ("C", "G"),
    ("E", "H"),  # Fixed missing comma above
}
G.add_edges_from(edge)

# 2. Run Breadth-First Search (BFS)
visited = []
parent = {}
queue = deque(["A"])

while queue:
    node = queue.popleft()
    if node not in visited:
        visited.append(node)
        for neighbour in sorted(G.neighbors(node)):
            if neighbour not in visited:
                parent[neighbour] = node
                queue.append(neighbour)

# 3. Setup Plot & Layout
pos = nx.spring_layout(G, seed=5)
fig, ax = plt.subplots(figsize=(10, 8))


# 4. Animation Function
def update(frame):
    ax.clear()

    # Track nodes visited up to the current frame
    visited_now = visited[: frame + 1]

    # Node coloring
    node_colors = [
        "red" if n in visited_now else "lightgray" for n in G.nodes()
    ]

    # Edge coloring based on parent relationships traversed so far
    edge_colors = []
    for u, v in G.edges():
        if (
            u in parent
            and parent[u] == v
            and u in visited_now
            and v in visited_now
        ) or (
            v in parent
            and parent[v] == u
            and u in visited_now
            and v in visited_now
        ):
            edge_colors.append("green")
        else:
            edge_colors.append("black")

    # Draw nodes and edges
    nx.draw_networkx_nodes(G, pos, ax=ax, node_color=node_colors, node_size=1000)
    nx.draw_networkx_labels(G, pos, ax=ax)
    nx.draw_networkx_edges(G, pos, ax=ax, edge_color=edge_colors, width=3)

    ax.set_title(f"Current Node: {visited[frame]}", fontsize=14)


# 5. Create & Display Animation
ani = FuncAnimation(
    fig, update, frames=len(visited), interval=1000, repeat=False
)

from IPython.display import HTML

HTML(ani.to_jshtml())
```
